and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
load_dotenv(override=True)

api_key = os.getenv("GEMINI_API_KEY")

openai = OpenAI(
    base_url="https://generativelanguage.googleapis.com/v1beta/openai",
    api_key=api_key
)
def get_response(question):
    response = openai.chat.completions.create(
        model="gemini-3.6-flash",
        messages=[
            {"role": "system", "content": """
                You are a technical tutor.

                Explain technical concepts clearly and simply.
                Start with a simple definition, explain how it works,
                and provide a practical example.
                """},
            {"role": "user", "content": question}
        ]
    )
    return response.choices[0].message.content
```

```python
# here is the question; type over this to ask something new

question = """
expain about the difference between LLMs and RAG
"""
print(display(Markdown(get_response(question))))
```
